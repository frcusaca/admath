This one answers a yes-or-no question. Python's words for yes and no are `True` and `False`.

```python
def is_home(spot):
    return np.allclose(spot, 0)      # is every number (very close to) zero?


print(is_home(np.array([0, 0])))
print(is_home(np.array([0, 1])))
print(is_home(1 * a + 1 * b - 1 * m))
```

## Part 3: Hunt for round trips with two moves

Use only `a` and `b`. Is there any round trip besides staying parked? Predict, then let the robot search.

`continue` means "skip this one and go on to the next".

```python
found = 0
for c in range(-6, 7):
    for d in range(-6, 7):
        if c == 0 and d == 0:
            continue                      # staying parked does not count
        if is_home(c * a + d * b):
            print("round trip!  c =", c, " d =", d)
            found = found + 1
print("round trips found:", found)
```

## The names for this

- If a set of moves has a round trip (not all weights zero), the moves are **linearly dependent**. At least one move is **redundant**: you could already do its job with the others.
- If the only way home is to stay parked, the moves are **linearly independent**. Every move adds something new.

## Deep 2: The direction counter

NumPy has a tool that counts how many **independent** directions a set of moves really has. It is called the rank. For now it is a black box.

Predict the count for each set, then run. Make up three more sets and predict those.

*This leads to: rank and solving systems (Module 9).*

```python
print(np.linalg.matrix_rank(np.column_stack([a, b])))
print(np.linalg.matrix_rank(np.column_stack([a, k])))
print(np.linalg.matrix_rank(np.column_stack([a, b, m])))
print(np.linalg.matrix_rank(np.column_stack([r, s, t])))
```

## Deep 3: Almost broken

Mode `k = (2, 4)` was useless next to `a = (1, 2)`. What if the repair shop gets it **almost** the same: `k2 = (2, 4.001)`?

Now `a` and `k2` are independent, so the rover can reach the charger at `(7, 4)`. Predict roughly how many minutes it takes. Then run.

What does that tell you about moves that are "nearly" dependent? What would happen with `(2, 4.000001)`?

*This leads to: numerical methods. Some problems are solvable on paper but fragile on a computer.*

```python
k2 = np.array([2, 4.001])
charger = np.array([7, 4])
weights = np.linalg.solve(np.column_stack([a, k2]), charger)
print(weights)
print(weights[0] * a + weights[1] * k2)
```

## Deep 4: Redundant features

Each kid below is described by three features. But two of the features are really the same information in different units.

Each **column** of the table is a vector with five numbers. How many independent directions do the three columns have? Predict, then run.

What does a redundant feature give your ML model? What if the inch measurements were rounded to whole inches?

*This leads to: why data scientists check their features before training.*

```python
height_cm = np.array([140.0, 152.0, 147.0, 160.0, 138.0])
height_in = height_cm / 2.54
age = np.array([10, 11, 10, 12, 10])

table = np.column_stack([height_cm, height_in, age])
print(table)
print("independent directions:", np.linalg.matrix_rank(table))
```

## Deep 6: A general round-trip hunter

Write a function `round_trips(moves, limit)` that takes a **list** of two moves and returns a list of all round trips with weights from `-limit` to `limit`. Then make it work for three moves.

```python
def round_trips(moves, limit):
    found = []
    for c in range(-limit, limit + 1):
        for d in range(-limit, limit + 1):
            if c == 0 and d == 0:
                continue
            if is_home(c * moves[0] + d * moves[1]):
                found.append((c, d))
    return found


print(round_trips([a, k], 3))
```
