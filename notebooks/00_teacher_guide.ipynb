## Conventions

**Notation.** $\langle a, b \rangle$ is the inner product. NumPy computes it with `np.dot(a, b)`, and Roblox with `a:Dot(b)`. Many books write $a \cdot b$ and say "dot product". $\lVert a \rVert$ is the length, and $\hat{a}$ is the length-1 version of $a$. For a number times a named vector we write $2a$. For a number times a written-out vector we write $2 \times (3, 4)$.

**Rows.** Vectors are written sideways. When vectors are stacked into a table in later modules, each vector is a row.

**Three movers.** The stories differ in what the mover is allowed to do. That difference is useful later, so keep it straight from the start.

| Mover | World | What it can do |
| --- | --- | --- |
| Car | 2D map, $(x, y)$, $y$ is north | Drives only along its heading. It must turn to change heading. |
| Rover (from Module 1) | 2D map | Has two fixed drive modes. It can run each one forward or in reverse for any time. |
| Roblox character | 3D, $(x, y, z)$, $y$ is up | Walks in any direction on the ground and can look a different way than it walks. NPCs may fly. |

## Part 0.0: What is a vector?

**The idea.** A vector is a list of numbers in which each slot always means the same thing. That is the whole definition, and it comes before any arrows. The kids meet nine examples side by side: a map spot, a Roblox position, a color, a game character, a pizza order, a snack bar, a day of weather, a calculator digit and a doctor's chart. Mixing them on purpose makes the point that a vector is not a picture. It is numbers plus an agreement about the slots.

Four ideas ride along, each in a sentence or two.

- **The order is a promise.** $(3, 4)$ is not $(4, 3)$. The meaning of each slot lives outside the vector, in a comment or in your head.
- **Dimension is the number of slots.** It has nothing to do with space. The Knight is a point in a 5-dimensional space of stats. Arrows are just what vectors with 2 or 3 slots look like.
- **What earns the name.** A list is worth calling a vector when adding two of them slot by slot, or scaling one, means something. Orders add. Recipes double. Phone numbers do neither.
- **Two warnings for later.** Slots may have different units, which matters once we measure length. And a slot may carry nothing new, as BMI does on the doctor's chart.

**Worked examples**

1. The Knight is $(9, 3, 8, 1, 5)$ for (strength, stealth, armor, magic, luck). Its armor is the third slot, 8. In NumPy that slot is `knight[2]`, because counting starts at 0.
2. $(255, 0, 0)$ is red and $(0, 0, 255)$ is blue. Same three numbers, different order, different color.
3. The digit 7 lights the top bar and the two right bars: $(1, 1, 1, 0, 0, 0, 0)$. The digit 1 is $(0, 1, 1, 0, 0, 0, 0)$ and the digit 8 is all ones.
4. Monday's pizza order $(2, 1, 0)$ plus Tuesday's $(1, 0, 3)$ is $(3, 1, 3)$. Double Monday's is $(4, 2, 0)$.
5. The doctor's chart $(40, 1.5, 17.8)$: BMI is weight divided by height squared, $40 \div 1.5^2 \approx 17.8$. Anyone holding the first two slots can fill in the third.

**Listen for.** "A vector is an arrow." Arrows are only the pictures for 2 or 3 slots. "Dimension means 3D." Ask how many dimensions the Knight has. "The third slot is `v[3]`." It is `v[2]`.

## Part 0.1: Vectors as moves

**The idea.** A vector is a list of numbers that travel together. On paper it is an arrow. A point is a place, and a vector is a move. The move from $P$ to $Q$ is $Q - P$, worked out slot by slot. The same move can start anywhere, so $(3, 4)$ starting at home and $(3, 4)$ starting at $(1, 1)$ are the same vector. In NumPy a vector is `np.array([3, 4])`, and `v[0]` is its first number.

**Point or vector?** The student notebook says this up front, because the same list of numbers is used both ways from the first page. $(1, 2, 3)$ read as a point is the tip of the vector $(1, 2, 3)$ drawn from home. The direction is always from $(0, 0, 0)$ to the point. Read this way, every point is also a vector, which is why one `np.array` serves for both.

Three rules cover almost every case: point $-$ point is a vector, point $+$ vector is a point, and vector $+$ vector is a vector. In Part 0.6, $P$ and $N$ are points, while $d$, $v$ and $f$ are vectors.

Most software stores the two the same way and leaves the reading to you. NumPy has only arrays. Roblox uses `Vector3` for both `Position` and velocity. A start point is asked for only when an arrow has to sit somewhere particular: `draw_vector(v, start=P)` here, matplotlib's `quiver(x, y, u, v)`, and a Roblox raycast, which takes an origin and a direction.

**Worked examples**

1. $P = (1, 2)$ and $Q = (4, 6)$. The move from $P$ to $Q$ is $(4 - 1, 6 - 2) = (3, 4)$. Check: start at $P$, go 3 east and 4 north, and you land on $Q$.
2. The printer nozzle is at $(120, 40, 0)$ and must go to $(90, 65, 0.2)$. The move is $(-30, 25, 0.2)$: back 30, over 25, up one layer.
3. In Roblox you stand at $(10, 0, 5)$ and an NPC stands at $(22, 0, 14)$. The move from you to the NPC is $(12, 0, 9)$. The middle number is 0 because you are both on the ground.

**Listen for.** "A vector is only a location." It can be read as one, by starting the arrow at home. But the same vector can start anywhere, and it is still the same vector. Also listen for kids who draw the arrow from the point back to home. The arrow goes from home to the point.

## Part 0.2: Adding, subtracting and scaling

**The idea.** Adding two vectors means making one move and then the other. On paper the second arrow starts at the tip of the first. In numbers, add element-wise (slot by slot); subtract element-wise too. Part 0.0 introduces the word *element-wise*: a hyphenated adjective meaning "one element (slot) at a time". Scaling by a number stretches the arrow and keeps its direction. A negative number flips it around. Subtracting is adding the opposite. Every slot is ordinary arithmetic, so the rules the kids already know still hold: order does not matter, and $2(a + b) = 2a + 2b$.

A velocity is a vector of "how far each second". Starting at $p$ with velocity $v$, the position after $t$ seconds is $p + tv$. This is the first linear combination the kids meet, though nobody says so yet.

**Worked examples**

1. $(3, 4) + (1, -2) = (4, 2)$. Draw $(3, 4)$, then draw $(1, -2)$ from its tip. The shortcut from home to the end is $(4, 2)$.
2. $2 \times (3, 4) = (6, 8)$. $-1 \times (3, 4) = (-3, -4)$. Half of $(3, 4)$ is $(1.5, 2)$.
3. A car starts at $(10, 0)$ with velocity $(3, 4)$ meters per second. After 5 seconds it is at $(10, 0) + 5 \times (3, 4) = (25, 20)$.
4. A Roblox character walks $(0, 0, 3)$ per frame on a conveyor that pushes $(1, 0, 0)$ per frame. Each frame it really moves $(1, 0, 3)$. After 4 frames it is at $(4, 0, 12)$.

**Listen for.** "The length of $a + b$ is the length of $a$ plus the length of $b$." Draw a triangle and measure. The shortcut is shorter unless the arrows point the same way.

## Part 0.3: Length and length-1 vectors

**The idea.** The length of a vector comes from the Pythagorean theorem: square each number, add, take the square root. In 3D there are three squares to add. Scaling a vector by 10 makes it 10 times as long. Dividing a vector by its own length gives $\hat{a}$, a vector of length 1 that points the same way. That keeps the direction and throws the size away. The distance between two points is the length of the move between them.

For a car, velocity $=$ speed $\times$ heading. The speed is the length of the velocity. The heading is its length-1 version.

Some vectors have whole-number lengths, which keeps paper work clean:

| Vector | Length |
| --- | --- |
| $(3, 4)$ | $5$ |
| $(5, 12)$ | $13$ |
| $(8, 15)$ | $17$ |
| $(7, 24)$ | $25$ |
| $(1, 2, 2)$ | $3$ |
| $(2, 3, 6)$ | $7$ |
| $(12, 15, 16)$ | $25$ |

**Worked examples**

1. $\lVert (3, 4) \rVert = \sqrt{9 + 16} = 5$. The length-1 version is $(0.6, 0.8)$. Check: $0.36 + 0.64 = 1$.
2. $\lVert (1, 2, 2) \rVert = \sqrt{1 + 4 + 4} = 3$. The length-1 version is $(\frac{1}{3}, \frac{2}{3}, \frac{2}{3})$.
3. A car's velocity is $(30, 40)$ kilometers per hour. Its speed is $50$ and its heading is $(0.6, 0.8)$.
4. A chaser must move 4 studs per frame toward a target whose move is $(12, 0, 9)$. The length is $15$, so the direction is $(0.8, 0, 0.6)$. The velocity is 4 times that: $(3.2, 0, 2.4)$.

**Listen for.** Kids who try to give $(0, 0)$ a length-1 version. It has no direction, so it has none.

## Part 0.4: The inner product: sign and size

**The idea.** To find $\langle a, b \rangle$, multiply the matching slots and add. The answer is a single number. That number carries two kinds of news.

- **Sign.** Positive means the angle is less than $90^\circ$. Zero means exactly $90^\circ$. Negative means more than $90^\circ$.
- **Size.** The number grows when either vector gets longer, and when the two line up better. Doubling $a$ doubles $\langle a, b \rangle$. Flipping $a$ flips the sign.

Zero is the special value. It is the only answer that does not change when the vectors are stretched. So "is it a right angle?" never needs square roots, while "what angle is it?" does. Part 0.5 picks that up.

A vector with itself gives its length squared: $\langle a, a \rangle = \lVert a \rVert^2$.

**Worked examples**

1. $\langle (3, 4), (4, -3) \rangle = 12 - 12 = 0$. The two arrows are perpendicular, though neither lies along an axis.
2. $\langle (3, 4), (3, 4) \rangle = 9 + 16 = 25 = 5^2$. Length is hiding inside the inner product.
3. A car points along $h = (3, 4)$. A station is at $d = (5, -10)$ from the car. $\langle h, d \rangle = 15 - 40 = -25$. Negative, so the station is behind. Nobody had to normalize $h$.
4. The car's left is $(-4, 3)$: swap the two numbers and negate the first. $\langle (-4, 3), (5, -10) \rangle = -20 - 30 = -50$. Negative, so the station is on the right. Two inner products place the station: behind and to the right.
5. You pull a sled with force $(40, 30, 0)$ newtons while it slides $(10, 0, 0)$ meters. The work done is $\langle \text{force}, \text{move} \rangle = 400$ joules. The 30 newtons sideways did nothing.

**Listen for.** Kids who write the answer as a vector, such as $(12, -12)$. Remind them: multiply, then add.

## Part 0.5: Raw or normalized?

**The idea.** The raw inner product mixes how long the vectors are with how well they line up. Dividing by both lengths removes the lengths. The result is the normalized inner product, and it always lands between $-1$ and $1$.

$$\frac{\langle a, b \rangle}{\lVert a \rVert \times \lVert b \rVert} = \langle \hat{a}, \hat{b} \rangle$$

The left side is Route B: inner product first, divide after. The right side is Route A: normalize first. They are equal because a number can be pulled out of either slot of an inner product. In the ML class this number is cosine similarity. It is the cosine of the angle, and `angle_between` turns it into degrees.

On paper Route B is kinder: the numbers stay whole until one division at the end. Route A is what code usually does, because the hat vectors can be stored and reused.

**Worked examples**

1. Compare $a = (3, 4)$ with four others.

| Other vector | Raw $\langle a, \cdot \rangle$ | Normalized |
| --- | --- | --- |
| $(6, 8)$ | $50$ | $1$ |
| $(24, 7)$ | $100$ | $0.8$ |
| $(-8, 6)$ | $0$ | $0$ |
| $(-6, -8)$ | $-50$ | $-1$ |

Raw says $(24, 7)$ is most like $a$. Normalized says $(6, 8)$ is, because it points exactly the same way.

2. Take $(3, 4)$ and $(8, 6)$. Route A: $(0.6, 0.8)$ and $(0.8, 0.6)$ give $0.48 + 0.48 = 0.96$. Route B: $48 \div (5 \times 10) = 0.96$.
3. Take $(5, 12)$ and $(12, 5)$. Route B gives $\frac{120}{169} \approx 0.7101$. Route A with the hat vectors rounded to 2 decimals gives $0.38 \times 0.92 + 0.92 \times 0.38 = 0.6992$. Rounding early cost almost two percent.

**Listen for.** "A bigger inner product means more alike." Example 1 is the answer. Also watch for vectors whose slots have different units, which is coin m0.5.10.

## Part 0.6: Roblox: toward, away and in view

**The idea.** You are at $P$ and an NPC is at $N$. Three vectors matter: the move $d = N - P$, your velocity $v$ and your facing $f$, which has length 1. Each of three questions uses a different amount of normalizing.

| Question | Work out | Normalize | Why |
| --- | --- | --- | --- |
| Am I getting closer? | $\langle v, d \rangle$ | nothing | Only the sign is needed. Dividing by lengths never changes a sign. |
| How fast am I closing in? | $\langle v, \hat{d} \rangle$ | $d$ only | The answer should be a speed, in studs per second. Normalizing $v$ too would throw the speed away. |
| Can I see it? | $\langle f, \hat{d} \rangle$ | both ($f$ already is) | Only directions matter. The view cone is $60^\circ$ to each side, and $60^\circ$ gives $0.5$. |

The rule of thumb for kids: **normalize exactly the things whose size you want to ignore.**

$\langle v, \hat{d} \rangle$ is the part of your velocity that points at the NPC. The rest of your speed goes sideways.

**Worked examples.** In all four, $P = (0, 0, 0)$ and $N = (6, 0, 8)$, so $d = (6, 0, 8)$, $\lVert d \rVert = 10$ and $\hat{d} = (0.6, 0, 0.8)$.

1. $v = (3, 0, 4)$. $\langle v, d \rangle = 18 + 32 = 50$, so you are getting closer. The closing speed is $\langle v, \hat{d} \rangle = 1.8 + 3.2 = 5$. Your speed is also 5, so you are walking straight at the NPC.
2. $v = (5, 0, 0)$. Same speed, 5. $\langle v, \hat{d} \rangle = 3$. You close in at 3 studs per second and slide sideways at 4. It is a 3-4-5 triangle.
3. $v = (-4, 0, 3)$. $\langle v, d \rangle = -24 + 24 = 0$. You are walking sideways around the NPC, neither closer nor farther for this instant.
4. $f = (1, 0, 0)$. $\langle f, \hat{d} \rangle = 0.6$, which is at least $0.5$, so the NPC is in view. The angle is about $53^\circ$.

**Listen for.** "I am facing it, so I am getting closer." Facing and walking are separate. Coin m0.6.8 has a character walking backwards toward the NPC. Also listen for "a bigger $\langle v, d \rangle$ means I am closing faster." That is only true while $d$ stays the same length, which is why the closing speed uses $\hat{d}$.

## Teacher's notes

**What a vector really is.** The grown-up definition agrees with the kid one. A vector is anything that can be added to its own kind and scaled by a number, with the usual rules. "A list where adding slot by slot means something" is that definition in fifth-grade words. Coin m0.0.9 is built on it.

**Three kinds of "nothing new".** Part 0.0 plants the idea of a redundant slot without sorting the kinds. For your own reference there are three, and the difference matters in Module 3.

| Example | How the extra slot is made | Linear algebra's verdict |
| --- | --- | --- |
| (math, reading, total) | total $=$ math $+$ reading | Linearly dependent. Module 3 catches it. |
| (Celsius, Fahrenheit) | $F = 1.8 \times C + 32$ | Dependent only once a column of ones is added. |
| (weight, height, BMI) | BMI $=$ weight $\div$ height$^2$ | Redundant, but not linearly dependent. Linear algebra does not see it. |

So BMI is a good hook and an honest warning: "can be worked out from the others" is a bigger idea than "linearly dependent".

**Why this is easy.** Vector addition follows the same five rules as adding whole numbers. Two vectors add to a vector. Order does not matter. Grouping does not matter. A zero vector changes nothing. Every vector has an opposite. Mathematicians call anything with these rules an abelian group. Kids have used the rules since grades 1 to 3 (Common Core 1.OA.B.3 and 3.OA.B.5). Do not use the name with them.

**Scaling adds one familiar rule, twice.** $2(a + b) = 2a + 2b$ and $(2 + 3)a = 2a + 3a$ are both the distributive property. Ask "where have you seen this before?" and let them say it is just like ordinary numbers. Coin m0.2.10 is built for this.

**The inner product is where the pattern breaks.** It takes two vectors and gives a number, not a vector. So three of them cannot be chained the way $2 \times 3 \times 4$ can. It still obeys two friendly rules: $\langle a, b \rangle = \langle b, a \rangle$, and a number can be pulled out of either slot, $\langle 2a, b \rangle = 2 \langle a, b \rangle$. Coin m0.4.8 lets kids find the second rule. That rule is the whole reason Route A equals Route B.

**Normalize what you want to ignore.** This is the sentence to repeat. Sign questions need no normalizing. "How much of this points along that" needs one hat. "How alike are these directions" needs two.

**Round late.** Coin m0.5.9 is a numerical-methods lesson in miniature. Route B keeps whole numbers until the last step, so it loses nothing.

**The units link.** An inner product adds the $x$-part to the $y$-part. That only makes sense when both parts have the same unit, or have been rescaled first. Coin m0.5.10 puts this to the kids, and the feature-vector unit returns to it.

**Why $0.5$.** The cosine of $60^\circ$ is $0.5$. Kids do not need the word cosine yet. "A $60^\circ$ angle gives $0.5$" is enough, and they can check it with `angle_between`.

**Where points and vectors really differ.** Sliding the whole world over by 5 studs changes every point and changes no vector: the move from the car to the station is the same wherever the map's home is. Graphics software marks the difference with one extra number, 1 for a point and 0 for a direction. Those are homogeneous coordinates, and the feature-vector unit lets the kids invent them.

**Roblox names.** `Magnitude`, `Unit` and `Dot` are the real Roblox Studio names, and a character's facing is its `LookVector`. Kids who script in Studio can try every Part 0.6 coin there.

**Python trouble to expect.** Counting starts at 0. Lists glue with `+` while arrays add. `=` stores a value and does not ask a question. `P[[0, 2]]` in Part 0.6 picks slots 0 and 2, which is new.

## Check the key

Run this cell. It recomputes every number in the answer key and stops with an error if one is wrong.

```python
import numpy as np

checked = 0


def same(got, want):
    """Stop with an error unless got and want are (almost) equal."""
    global checked
    assert np.allclose(np.array(got, dtype=float), np.array(want, dtype=float), atol=0.005), (got, want)
    checked = checked + 1


def v(*numbers):
    return np.array(numbers, dtype=float)


def length(a):
    return np.linalg.norm(a)


def hat(a):
    return a / np.linalg.norm(a)


def inner(a, b):
    return float(np.dot(a, b))


def normalized(a, b):
    return inner(a, b) / (length(a) * length(b))


# Part 0.0
same(v(2, 1, 0) + v(1, 0, 3), (3, 1, 3))
same(2 * v(2, 1, 0), (4, 2, 0))
same(2 * v(2, 3, 1), (4, 6, 2))
same(round(40 / 1.5 ** 2, 1), 17.8)
digits = {0: "1111110", 1: "0110000", 7: "1110000", 8: "1111111"}       # top, upper right, lower right, bottom, lower left, upper left, middle
assert digits[7] == "1110000" and digits[1] == "0110000" and digits[8] == "1" * 7 and digits[0] == "1111110"

# Part 0.1
same(v(1, 1) + v(4, 3), (5, 4))
same(v(5, 5) - v(2, 1), (3, 4))
same(v(2, 1) - v(5, 5), (-3, -4))
same(v(22, 0, 14) - v(10, 0, 5), (12, 0, 9))
same(v(120, 40, 0) + v(-30, 25, 0.2), (90, 65, 0.2))
same(v(112, 12, -188) - v(100, 64, -200), (12, -52, 12))
same(v(3, 2) - v(3, 7), (0, -5))
same(v(10, 4) - v(6, -2), (4, 6))
same(v(4, 5) - v(1, 1), (3, 4))
same(v(7, 9) - v(4, 5), (3, 4))

# Part 0.2
same(v(2, 5) + v(4, 1), (6, 6))
same(v(2, 5) - v(4, 1), (-2, 4))
same(3 * v(2, -1), (6, -3))
same(-2 * v(1, 4), (-2, -8))
same(v(1, 2, 3) + v(10, 20, 30), (11, 22, 33))
same(2 * v(1, 3) + 3 * v(2, -1), (8, 3))
same(v(10, 0) + 5 * v(3, 4), (25, 20))
same(4 * (v(0, 0, 3) + v(1, 0, 0)), (4, 0, 12))
same(v(7, 1) - v(2, 5), (5, -4))
same(2 * (v(3, 4) + v(1, -2)), (8, 4))
same(2 * v(3, 4) + 2 * v(1, -2), (8, 4))

# Part 0.3
same(length(v(6, 8)), 10)
same(length(v(5, 12)), 13)
same(length(v(-3, 4)), 5)
same(length(v(2, 3, 6)), 7)
same(hat(v(0, -7)), (0, -1))
same(hat(v(8, 6)), (0.8, 0.6))
same(length(v(22, 0, 14) - v(10, 0, 5)), 15)
same(length(v(30, 40)), 50)
same(hat(v(30, 40)), (0.6, 0.8))
same(4 * hat(v(12, 0, 9)), (3.2, 0, 2.4))
same([length(v(30, 40)), length(v(-6, -8)), length(v(1.5, 2))], [50, 10, 2.5])

# Part 0.4
same(inner(v(2, 3), v(4, 1)), 11)
same(inner(v(1, 5), v(5, -1)), 0)
same(inner(v(2, 1), v(-3, 1)), -5)
same(inner(v(1, 2, 3), v(4, 0, -1)), 1)
same(inner(v(6, 8), v(6, 8)), 100)
same(inner(v(3, 4), v(5, -10)), -25)
same(inner(v(40, 30, 0), v(10, 0, 0)), 400)
a, b = v(3, 4), v(2, 1)
same([inner(2 * a, b), inner(a, 3 * b), inner(2 * a, 3 * b), inner(-a, b)], [20, 30, 60, -10])
same(inner(v(3, -2), v(4, 6)), 0)
same(inner(v(-4, 3), v(3, 4)), 0)          # (-4, 3) really is perpendicular to the heading
same(inner(v(-4, 3), v(5, -10)), -50)

# Part 0.5
same(normalized(v(3, 4), v(6, 8)), 1)
same(normalized(v(3, 4), v(-8, 6)), 0)
same(normalized(v(3, 4), v(-6, -8)), -1)
same(normalized(v(3, 4), v(24, 7)), 0.8)
same(inner(hat(v(20, 15)), hat(v(7, 24))), 0.8)
same(inner(v(20, 15), v(7, 24)) / (25 * 25), 0.8)
knight, rogue, mage = v(9, 3, 8, 1, 5), v(4, 9, 3, 2, 7), v(2, 5, 1, 9, 6)
same([inner(knight, rogue), inner(knight, mage), inner(rogue, mage)], [124, 80, 116])
same([normalized(knight, rogue), normalized(knight, mage), normalized(rogue, mage)], [0.73, 0.49, 0.76])
same(normalized(v(5, 12), v(12, 5)), 120 / 169)
same(120 / 169, 0.7101)
same(inner(np.round(hat(v(5, 12)), 2), np.round(hat(v(12, 5)), 2)), 0.6992)

# Part 0.6
d = v(6, 0, 8)
same(length(d), 10)
same(inner(v(3, 0, 4), d), 50)
same(inner(v(-4, 0, 3), d), 0)
same(inner(v(0, 0, -5), d), -40)
same(inner(v(1, 0, 0), d), 6)
same([inner(v(3, 0, 4), hat(d)), inner(v(5, 0, 0), hat(d))], [5, 3])
same([inner(v(1, 0, 0), hat(d)), inner(v(0, 0, 1), hat(d)), inner(v(0, 0, -1), hat(d))], [0.6, 0.8, -0.8])
same(inner(v(-0.6, 0, -0.8), hat(d)), -1)
same([length(v(2, 3, 6)), length(v(2, 6, 3))], [7, 7])
same([inner(v(0, 0, 1), hat(v(2, 3, 6))), inner(v(0, 0, 1), hat(v(2, 6, 3)))], [6 / 7, 3 / 7])
f = v(0, 0, 1)
same([inner(f, hat(v(3, 0, 4))), inner(f, hat(v(8, 0, -6))), inner(f, hat(v(-12, 0, 5)))], [0.8, -0.6, 5 / 13])

# Proficiency check
same(v(2, -1) + v(1, 4), (3, 3))
same(length(v(5, 12)), 13)
same(inner(v(2, -1), v(1, 2)), 0)
same(inner(v(-3, 0, -4), v(6, 0, 8)), -50)

print("All", checked, "checks passed.")
```
